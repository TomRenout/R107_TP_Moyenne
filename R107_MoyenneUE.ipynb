```python
#Algorithme du programme :
#on demande a l'utilisateur chaque note pour chaque ressource et SAE.
#on fait le calcul des moyennes à l'aide des coefficients pour chaque ressources et SAE dans chaque UE.
#pour les récupérer, on demande a l'utilisateur la note de sa première ressource, ensuite on lui redemande grâce à une variable liste, si c'est 1 et/ou
#2 et/ou 3, et pour chaque UE choisis on demande le coefficient pour chaque UE.
#une fois qu'on à chaque moyenne d'UE, on les compare, savoir si au moins deux sans au minimum de 10, et s'il y en a que deux, vérifié que la 3eme
#soit bien au dessus de 8.
```

```python
def TPmoy():
    totcoeff1=0
    totcoeff2=0
    totcoeff3=0
    totue1=0
    totue2=0
    totue3=0
    compteur=1
    coeffue=0
    coeffue1=0
    coeffue2=0
    autreue1=0
    autreue2=0
    for i in range(101,103):
        note=float(input("Veuillez donnez votre note de la R{}".format(i)))
        quelue=float(input("Quel UE ? (1, 2, 3)"))
        while quelue != 1 and quelue !=2 and quelue !=3:
            quelue=float(input("Quel UE ? (1, 2, 3)"))
            
        coeffue+=float(input("Donnez votre coefficient d'UE"))
        autreue=input(f"D'autre UE compte pour la R{i} ? Oui (o) Non (n)")
        while autreue.lower() != "o" and autreue.lower() != "n":
            autreue=input(f"D'autre UE compte pour la R{i} ? Oui (o) Non (n)")
            
        if autreue.lower() == "o":
            compteur+=1
            autreue1=float(input("Quel autre UE compte ?"))
            coeffue1+=float(input("Donnez votre coefficient d'UE"))
            autreue=input(f"D'autre UE compte pour la R{i} ? Oui (o) Non (n)")
            
            while autreue.lower() != "o" and autreue.lower() != "n":
                autreue=input(f"D'autre UE compte pour la R{i} ? Oui (o) Non (n)")
                
            if autreue.lower() == "o":
                compteur+=1
                autreue2=float(input("Quel autre UE compte ?"))
                coeffue2=float(input("Donnez votre coefficient d'UE"))
                
        for k in range(0, compteur):
            if quelue == 1:
                totue1+=note*coeffue
                quelue=0
            elif autreue1==1:
                totue1+=note*coeffue1
                autreue1=0
            elif autreue2==1:
                totue2+=note*coeffue2
                autreue2=0
            elif quelue ==2:
                totue2+=note*coeffue
                quelue=0
            elif autreue1==2:
                totue2+=note*coeffue1
                autreue1=0
            elif autreue2==2:
                totue2+=note*coeffue2
                autreue2=0
            elif quelue == 3:
                totue3+=note*coeffue
                quelue=0
            elif autreue1 == 3:
                totue3+=note*coeffue1
                autreue1=0
            elif autreue2==3:
                totue3+=note*coeffue2
                autreue2=0
                
    print("Votre moyenne pour l'UE1 est :"+float(totue1))
    print("Votre moyenne pour l'UE2 est :"+float(totue2))
    print("Votre moyenne pour l'UE3 est :"+float(totue3))
TPmoy()
```
