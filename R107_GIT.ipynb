## 01- Notion de variable


**Définition**

Une **variable** est un nom qui permet de repérer une zone de la mémoire centrale dans laquelle une valeur est stockée. Elle correspond donc à une adresse (on pourrait l'assimiler à une boîte aux lettres) à laquelle le programmeur accède grâce à son nom.

Une variable fait donc le lien entre le nom de la variable que l'on manipule dans le programme et une adresse où est physiquement stockée la donnée.

Les données manipulées par le programme sont stockées le temps de son exécution en mémoire centrale. Les variables permettent de manipuler ces données sans avoir à se préoccuper de l'adresse explicite qu'elles occuperont effectivement en mémoire. Ceci n'est possible que parce que l'interpréteur attribue à notre insu une adresse à chaque variable et supervise tous les changements d'adresse qui pourraient s'en suivre.

*Exemple*

"a = 5" signifie "à l'adresse mémoire référencée par "a" (ie dans la boîte aux lettres nommée "a") se trouve la valeur 5".

**Remarque**

En informatique, le signe "=" signifie l'**affectation** ou l'**assignation** d'une valeur à une variable.

*Règles de syntaxe*

* Le nom d’une variable ne peut être composé que de lettres, majuscules ou minuscules, de chiffres et du symbole « _ » (*underscore* en anglais). Il ne peut y avoir d’espace dans le nom d’une variable.

* Le nom d’une variable ne peut pas commencer par un chiffre.

* Le langage Python est sensible à la casse, ce qui signifie que Python fait la différence entre les majuscules et les minuscules.

*Les conventions des programmeurs*

* Noms des variables écrits en minuscules, les espaces éventuels étant remplacés par des « _ ».

  *Exemple : annee_de_naissance*

* Noms des variables écrits en minuscules mais premier caractère de chaque mot en majuscule, sauf le premier.

  *Exemple : anneeDeNaissance*

## 02 - Structures de contrôle

L'interpréteur Python parcourt le code de façon séquentielle. Il est parfois nécessaire de modifier la séquence naturelle d'exécution en bifurquant vers une nouvelle instruction qui n'est plus la suivante dans la séquence naturelle.  
Il existe deux formes d'instructions permettant de rompre avec la séquence en cours : la forme **conditionnelle** et la forme **répétitive**.

### 📚 1. Instructions conditionnelles

Elles permettent au code de suivre différents chemins selon les circonstances.

**Syntaxe**

### 📚 2. Structures itératives

Elles permettent de répéter certaine opération autant de fois que nécessaire.

#### 📌 La boucle while

Elle permet de répéter un bloc d'instructions **tant qu'une condition est vraie**.

**Syntaxe**

Dès que la condition cesse d'être vérifiée, le bloc "subordonné" au **while** est ignoré et c'est l'instruction suivante du code qui est exécutée.  

❗️_Points de vigilance_
* Les variables intervenant dans la condition doivent être initialisées avant toute évaluation dans la boucle
* Une condition toujours vérifiée se traduira par une "boucle infinie" : il faut donc veiller à faire évoluer dans le bloc au moins une des variables intervenant dans l'écriture de la condition.

#### 📌 La boucle for

Elle permet d'itérer sur une collection de données, telle une liste ou un dictionnaire.

**Exemples**

```python
# Exemple d'utilisation d'une boucle for sur une liste

liste=['Alice','Boualem','Carla','Djaid','Ewen','Farid']  #les mots d'une liste sont placés entre crochets

# boucle permettant l'affichage de tous les éléments de la liste
for element in liste: # pour chaque élément de la liste
    print(element) # afficher l'élément

# boucle permettant l'affichage de tous les caractères de l'élément placé à la position 3 dans la liste
for x in liste[3]: # attention ! la numérotation des éléments de la liste commence à zéro !
    print(x)
```

```python
# Exemple d'utilisation d'une boucle for avec un compteur

liste=['Alice','Boualem','Carla','Djaid','Ewen','Farid']
for i in range(len(liste)): # signifie "pour i dans l'intervalle allant de 0 à la longueur de la liste -1
    print(liste[i]) # afficher l'élément numéro i de la liste
```

❗️**Remarque importante**

* **len(liste)** est la longueur de la liste, soit encore le nombre d'éléments dans la liste
* Le compteur commence à 0 par défaut et la boucle est exécutée sur l'ensemble des éléments de la liste soit len(liste) fois. Le compteur s'arrête donc à la valeur de len(liste)-1.

## 03- Fonctions et modules

En programmation, on est souvent amené à réutiliser plusieurs fois une même suite d'instructions. Les **fonctions** vont nous permettre de regrouper ces instructions dans un bloc qui sera "appelé" grâce à un nom.

Les **modules** permettront, quant à eux, de regrouper plusieurs fonctions selon le même principe. À titre d'exemple, les fonctions mathématiques peuvent être regroupées dans un module dédié.

### 📚 1. Construction d'une fonction

**Remarques**
 * Le mot-clé **def** est l'abréviation de "_define_".
 * Dans la liste des paramètres qui seront fournis lors d'un appel à la fonction, les espaces sont optionnels mais ils améliorent la lisibilité du code.

**Exemple 1 :** définir une fonction nommée "exemple_1" qui affiche les entiers compris entre 1 et 8. On notera qu'ici aucun paramètre n'est à indiquer (la liste des paramètres est donc vide).

```python
# Définition de la fonction exemple_1 qui affiche les entiers compris entre 1 et 8

def exemple_1():
    # boucle signifiant "pour i allant de 1 à 9-1, afficher la valeur de i"
    for i in range(1,9):        # l'indentation est nécessaire
        print(i)
# La dernière exécution de la boucle correspond à la valeur 8
```

❗️L'exécution de ces instructions correspond uniquement à la construction de la fonction et elle ne donne lieu à aucun affichage de valeurs de i. Pour obtenir l'affichage des valeurs, il faut **appeler la fonction** (la faire fonctionner en somme).

On notera également que le symbole **#** permet d'écrire des **commentaires** dans les lignes de code. Rien de ce qui apparaît en italique après le symbole ne sera exécuté.

```python
# Instruction permettant d'appeler une fonction
exemple_1()
```

**Exemple 2 :** définir une fonction nommée "exemple_2" qui affiche les entiers pairs compris entre deux entiers positifs _a_ et _b_.

```python
# Définition de la fonction exemple_2 qui affiche les entiers pairs compris entre deux entiers positifs a et b

def exemple_2(a,b):  # ici on définit les deux entiers positifs a et b comme des paramètres de la fonction
    # boucle signifiant "pour i allant de a à b (b+1-1 !), si i est divisible par 2 alors afficher la valeur de i"
    for i in range(a,b+1):
        if i%2==0:   # on effectue la division euclidienne de i par 2 et on vérifie si le reste est égal à 0
            print(i)

# appel de la fonction en prenant pour valeurs a=1 et b=20 (par exemple !)
exemple_2(1,20)
```

❗️ Lorsqu'on effectue un test de comparaison entre deux valeurs, ici : **if i%2==0** (signifie précisément "si le reste dans la division euclidienne de i par 2 est égal à 0"), il faut veiller à **doubler le signe "="**.

L'écriture i%2=0 signifie effectivement que l'on affecte la valeur 0 à une variable nommée i%2, ce qui n'est pas possible d'une part (voir les règles de nommage des variables), et ne correspond pas à ce que l'on souhaite faire d'autre part. Le fait de doubler le signe "=" indique que l'on **compare** les deux valeurs.

**Bon à savoir : les opérateurs de comparaison**

| Opérateur| Signification littérale|
| :--------|:---------------|
| <        |strictement inférieur à|
| >        |strictement supérieur à|
| <=       |inférieur ou égal à|
| >=       |supérieur ou égal à|
| ==       |égal à|
| !=       |différent de|

### 📚 2. Importation de modules ou de fonctions contenues dans un module

##### Un peu d'exploration...

En anglais, "racine carrée" se dit _"square root"_ et la fonction mathématique qui permet de la calculer est "sqrt" mais ...

```python
sqrt(9)
```

La fonction **sqrt** n'est pas disponible dans l'espace de noms principal d'où le message d'erreur renvoyé par l'exécution du code. C'est une fonction que l'on trouve par exemple dans le module **math** et il est nécessaire de l'importer depuis ce module pour être en mesure de l'utiliser. 

```python
# on importe la fonction sqrt depuis le module math
from math import sqrt
```

```python
sqrt(9)
```

La fonction, ainsi importée depuis le module **math**, devient accessible et est désormais définie dans l'espace de noms principal.

Le module math rassemble un très grand nombre de fonctions mathématiques. On y trouve également la définition de valeurs importantes telles que $\pi$ ou $e$.

On peut consulter l'aide sur ce module à l'aide de l'instruction :

```python
# affichage de l'aide relative au module math
help("math")
```

Si on veut exploiter la valeur $\pi$ :

```python
from math import pi
print(pi)
```

Si on souhaite disposer de l'ensemble des fonctions définies dans le module on peut importer le module dans sa totalité.

```python
# on importe le module math complet
import math
e
# la valeur de e n'est pas accessible directement malgré l'importation du module complet
```

```python
# syntaxe particulière permettant d'appeler la valeur e contenue dans le module math
math.e
```

##### Bilan quant à l'importation de modules

###### Première méthode

Python importe la totalité du module **math**. Toutes les fonctions sont accessibles et on les appelle par leur nom précédé du nom du module et d'un point.

Exemple : math.sqrt(16) renvoie 4.0

Dans ce cas, Python crée un espace de noms "math" contenant les variables et les fonctions du module math, autre que l'espace de noms principal dans lequel on travaille.

###### Deuxième méthode

Python charge la fonction depuis le module math et la place dans l'interpréteur au même plan que les fonctions déjà existantes dans l'espace de noms principal.

Exemple : ici sqrt(16) renverra 4.0 mais pi ne sera pas reconnue.

**Remarque :** on peut aussi importer l'ensemble des fonctions et variables selon cette deuxième méthode, en utilisant la syntaxe suivante :

Dans ce cas, l'ensemble des fonctions et variables sont placées dans l'interpréteur au même plan que les fonctions qui étaient déjà présentes.

* Écrire une fonction permettant de déterminer le nombre de solutions réelles d'une équation du second degré.

**Rappel : résolution d'une équation du second degré de type $ax^2+bx+c=0$, calcul du discriminant**

\begin{equation}
\Delta = b^2 - 4 ac
\end{equation}  

Si $\Delta>0$, alors l'équation admet deux solutions réelles distinctes :
\begin{equation}
\begin{cases}
x_1 = \dfrac{-b - \sqrt{\Delta}}{2a}\\
x_2 = \dfrac{-b + \sqrt{\Delta}}{2a}
\end{cases}
\end{equation}  

Si $\Delta = 0$, alors l'équation admet une solution réelle double :
\begin{equation}
x = \dfrac{-b}{2a}
\end{equation}  

Si $\Delta<0$, alors l'équation n'admet aucune solution réelle
